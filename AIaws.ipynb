# Class method decorator

```python
from datetime import date
```

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year-birthyear)
    def show(self):
        print(self.name+"'s age is "+str(self.age))
```

```python
age=student("Greeshma",21)
age.show()
```

```python
by=student.calculate_age('Greeshma',2005)
by.show()
```

# Mini ATM project

```python
class BankAccount:
    def __init__(self,name,balance=0):
        #instance attributes
        self.name=name
        self.balance=balance
    #deposit money
    def deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print("deposit Success")
            print("deposited:",amount)
        else:
            print("Invalid deposit amount")
    #withdraw money
    def withdraw(self,amount):
        if amount<=0:
            print("invalid withdrawal amount")
        elif amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
            print("Withdrawl successful")
            print("withdrawn amount:",amount)
    #check balance
    def check_balance(self):
        print("Current balance:", self.balance)
    #display account details
    def display_details(self):
        print("account holder:",self.name)
        print("balance:",self.balance)
```

```python
acc1=BankAccount("Greeshma",5000)
acc1.display_details()
```

```python
acc1.deposit(30000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(2000)
```

# Inheritance

```python
#parent class
class Person:
    def person_info(self,name,age):
        print('Inside Person Class')
        print('Name: ',name,'Age: ',age)
#parent class 2
class Company:
    def company_info(self,company_name,location):
        print('Inside the company class ')
        print('Name:',company_name, 'Location:',location)
        
```

```python
#child class
class Employee(Person,Company):
    def Employee_info(self,salary,skill):
        print('Inside employee class - child class')
        print('Salary',salary,'Skills:',skill)
```

```python
#create an object of employee
emp2=Employee()
```

```python
#access the data
emp2.person_info('Greeshma',21)
emp2.company_info('MSRG Foundation','Bengaluru')
emp2.Employee_info(50000,'Forward deployed Engineer')
```

# Access Specifiers

```python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    # public instance method
    def show(self):
        #access public data members
        print('Name',self.name,'Salary',self.salary)
        
```

```python
emp=Employee('Greeshma',10000)
#access the public data members
print('Name',emp.name,'Salary',emp.salary)
emp.show()
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name=name # public data member
        self.__salary=salary # private data member 
```

```python
#create an object
emp=Employee("Greeshma",50000)
#accessing private data member
print('salary:',emp.__salary)  # we cannot access a private data directly. So it is trowing the error
```

```python
class Company:
    def __init__(self):
        #protected member
        self._project="ML project"
#child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("Employee name:",self.name)
        #access the protected member
        print("Working on project:",self._project)
    
```

```python
c=Employee("Greeshma")
c.show()
```

```python
# direct access
print('Project:',c._project)
```

```python
class India():
    def capital(self):
        print("New Delhi is the capital of India")
    def language(self):
        print("India has 22 official languages")
    def type(self):
        print("India is a developing nation!")
class USA():
    def capital(self):
        print("Capital of USA is Washington DC")
    def language(self):
        print("The official language is US is English")
    def type(self):
        print("USA is a developed nation!")
```

```python
class Bird:
    def info(self):
        print("There are many types of birds!")
    def flight(self):
        print("Most of the birds can fly but some cannot!")
class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly!")
class ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly!")
```

```python
ob_bird=Bird()
ob_bird.info()
ob_bird.flight()
```

```python
ob_ostrich=ostrich()
ob_ostrich.flight()
```
