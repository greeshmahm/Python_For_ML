# Comments

```python
# Get the name of the file and open it
name=input('Enter file:')
handle=open(name,'r')

#Count wor frequency
counts=dict()
for line in handle:
    words=line.split()
    for word in words:
        counts[word]=counts.get(word,0)+1

# Find the most common word
bigcount=None
bigword=None
for word,count in counts.item():
    if bigcount = word
    bigcount=count

# All done
print(bigword,bigcount)
```

# Comparison operators

```python
x=5
if x==5:
    print('Equals 6')
if x>4:
    print('Greater than 4')
if x>=5:
    print('Greater than or equal 5')
if x<=6:
    print('Less than 6')
if x<=5:
    print('Less than or equals 5')
if x!=6:
    print('Not equal 6')
```

```python
x=5
print('Before 5')
if x==5:
    print('Is 5')
    print('Is still 5')
    print('Third 5')
print('Afterwards 5')
print('Before 6')
if x==6:  # 5!=6 =>false so the below 3 set of print statements of this 'if' does not get executed.(CONDITIONAL CODE)
    print('Is 6')
    print('Is Still 6')
    print('Third 6')
print('Afterwards 6') # out of 'if' statement so it get's executed.(SEQUENCIAL CODE)
```

```python
x=5
if x>=2:
    print('Bigger than 2')
    print('Still bigger')
print('Done with 2')
# Nested loops(loop inside loop)
for i in range(5):
    print(i)
    if i>2:
        print('Bigger than 2')
    print('Done with i =',i)
print('All done')
```

```python
# nested if
x=42
if x>1:
    print('More than one')
    if x<100:
        print('Less than 99')
print('All done')
```

# Two-way Decisions -with else: if not this do this

```python
x=4
if x>2:
    print('Bigger')
else:
    print('Smaller')
print('All done')
```

# Multi-way

```python
x=5
if x<2:
    print('Small size')
elif x<10:
    print('Medium size')
else:
    print('Large')
print('ALL IS WELL')
```
