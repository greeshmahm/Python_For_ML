```python
# Get the name of the file and open it
name=input('Enter file:')
handle=open(name,'r')

#Count wor frequency
counts=dict()
for line in handle:
    words=line.split()
    for word in words:
        counts[word]=counts.get(word,0)+1

# Find the most common word
bigcount=None
bigword=None
for word,count in counts.item():
    if bigcount = word
    bigcount=count

# All done
print(bigword,bigcount)
```

```python
x=5
print('Before 5')
if x==5:
    print('Is 5')
    print('Is still 5')
    print('Third 5')
print('Afterwards 5')
print('Before 6')
if x==6:  # 5!=6 =>false so the below 3 set of print statements of this 'if' does not get executed.(CONDITIONAL CODE)
    print('Is 6')
    print('Is Still 6')
    print('Third 6')
print('Afterwards 6') # out of 'if' statement so it get's executed.(SEQUENCIAL CODE)
```

```python
'''Write a program to prompt for a score b/w 0.0 and 1.0. If the score is out of range,print an error.
   If the score is b/w 0.0 and 1.0, print a grade using following table:
   Score grade
   >=0.9A
   >=0.8B
   >=0.7C
   >=0.6D
   <0.6F
   If the user enters a value out of range, print a suitable error message and exit.For the test, enter a score of 0.85.
'''
score=input('Enter Score:')
try:
    score=float(score)
except:
    print("Bad score")
    quit()
if score<0.0 or score>1.0:
    print("Bad score")
    quit()
if score>=0.9:
    print("A")
elif score>=0.8:
    print("B")
elif score>=0.7:
    print("C")
elif score>=0.6:
    print("D")
else:
    print('F')
```

A parameter is a variable which we use in the function definition. It is a "handle" that allows the code in the function to access the arguments for a particular function invocation.

```python
def greet(lang):
    if lang=='es':
        print('Hola')
    elif lang=='fr':
        print('Bonjour')
    else:
        print('Hello')
greet('en')
```

# Parameters and Arguments

```python
# 'name' and 'age' are PARAMETERS(placeholders)
def greet(name,age):
    print(f"Hello {name}, you are {age} years old.")
# "Greeshma" and 21 are ARGUMENTS(the real values)
greet("Greeshma",21)
```
