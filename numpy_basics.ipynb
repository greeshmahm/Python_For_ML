# What is NumPy

```python
#  numpy is a multidimensional array which is used to store all kinds of data in 1D, 2D, 3D arrays.
# Why numpy over lists
# list is slow compared to numpy.
# numpy uses fixed type. ex: 5-> 00000101     -> numpy takes: int 32 space to store binary 5.
#                                             -> list   takes a lot of space than numpy.
# ie., Why is NumPy faster?
#         * Fixed type of data.                                       
#         * Faster to read less bytes of memory.
#         * No type checking when iterating through objects.
#         * Numpy uses contiguous memory like array.(list is a dynamic array: slow bcz the fetching process is slow compared to numpy)
#         Numpy: (start, total_size,type of memory block) 
#         * Numpy can use SIMD vector Processing: Single instruction multiple data
#        So if we want to do any addition of lot of values, instead of doing one addition at a time 
#         we can use this SIMD vector and perform the computation on all these values at one time.
#        Numpy effectively uses Cache memory to load/store the values(while performing any operations) which is close to CPU 
#        Both numpy and lists operations: Insertion, deletion,appending, concatenation,etc.
#        ex: Numpy allows as to do single valued computation, but lists does not allow.
#              lists   a[1,3,5]
#                      b[1,2,3]
#                      a*b = ERROR
#              NumPy   a=np.array([1,3,5])
#                      b=np.array([1,2,3])
#                      a*b=np.array([1,6,15]) # 1*1=1, 3*2=6, 5*3=15
# # Applications of numpy:
#   * NumPy can do all sorts of Mathematics (So NumPy can be used in place of MATHLAB).
#   * SciPy library has even more Mathematics than NumPy.SciPy has more functions than NumPy 
#   * Numpy can be used in Plotting (Matplotlib)
#   * it is Backend of different applications like Pandas, Connect 4, Digital Photography.
#   * Useful in Machine Learning applications both directly and indirectly bcz we use Tensors in ML which is more similar to NumPy, Tensors = way to store all sorts of values.
```

#### Be careful when copying arrays!!

```python
a=np.array([1,2,3])
b=a
print(a)
```

```python
a=np.array([1,2,3])
b=a       # this does not makes the copy of the array so it is pointing to the same thing.
b[0]=25   
print(a)
```

```python
a=np.array([1,2,3])
b=a.copy()   # it does not modify the existing array
b[0]=25
print(a)
```

# Statistics

```python
stats=np.array([[1,2,3],[4,5,6]])
stats
```

```python
np.max(stats)
```

```python
np.min(stats, axis=1) # axis=1 => gives the min value in the first and second row
```

```python
np.min(stats)
```

```python
np.sum(stats)  #1+2+3+4+5+6=21
```

```python
np.sum(stats,axis=0) # 1+4=5, 2+5=7, 3+6=9 
```

```python
np.sum(stats,axis=1)  # adds each row values ie., 1+2+3=6, 4+5+6=15
```

# Reorganizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)
print(before.shape) # 2rows and 4 col.
```

```python
after=before.reshape((8,1))
print(after)
```

```python
after=before.reshape((4,2))
print(after)
```

```python
# Vertically stacking vectors
v1=np.array([1,2,3,4])
v2=np.array([5,6,7,8])
np.vstack([v1,v2,v2,v2,v2])
```

```python
# Horizontal stack
h1=np.ones((2,4))
h2=np.zeros((2,2))
np.hstack((h1,h2,h1,h1,h2))
```

# Boolean Masking and Advanced Indexing

```python
filedata>50
```

```python
np.all(filedata>50,axis=0)
```

```python
filedata[filedata>50]  # which are the values that are greater than 50.
```

```python
((filedata>50) & (filedata<100))
```

```python
(~((filedata>50) & (filedata<100)))
```
